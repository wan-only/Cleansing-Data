# 2. LOAD DATASET

```python
file_path = "/content/drive/MyDrive/backup/dataset/dataset_30_baris_dirty_asli.csv"
```

```python
df_dirty = pd.read_csv(file_path, sep=";")
print("=== Dataset Asli ===")
print(df_dirty.head())
print("\nInfo Dataset:")
print(df_dirty.info())
print("\nJumlah Data & Kolom:")
print(df_dirty.shape)
```

# 3.CEK MISSING VALUE

```python
print("\n=== CEK MISSING VALUES SEBELUM CLEANING ===")
missing_before = df_dirty.isnull().sum()
print(missing_before)

print("\nJumlah total missing value:",
      df_dirty.isnull().sum().sum())
```

# 4.PERBAIKAN MISSING VALUE

```python
print("\n=== PROSES PERBAIKAN MISSING VALUES ===")

# ------------------------------------------------------------
# total_qty → isi dengan median
# ------------------------------------------------------------
if df_dirty['total_qty'].isnull().sum() > 0:
    median_qty = df_dirty['total_qty'].median()

    print(f"Missing total_qty ditemukan: {df_dirty['total_qty'].isnull().sum()}")
    print(f"Median total_qty: {median_qty}")

    df_dirty['total_qty'] = df_dirty['total_qty'].fillna(median_qty)

    print("total_qty berhasil diisi dengan median.")
else:
    print("total_qty tidak memiliki missing value.")


# ------------------------------------------------------------
# Metode Pembayaran → isi Unknown
# ------------------------------------------------------------
if df_dirty['Metode Pembayaran'].isnull().sum() > 0:

    jumlah_missing = df_dirty['Metode Pembayaran'].isnull().sum()

    print(f"Missing Metode Pembayaran ditemukan: {jumlah_missing}")

    df_dirty['Metode Pembayaran'] = (
        df_dirty['Metode Pembayaran']
        .fillna("Unknown")
    )

    print("Metode Pembayaran berhasil diisi dengan 'Unknown'.")
else:
    print("Metode Pembayaran tidak memiliki missing value.")
```

# 5. Cek ulang Missing Values

```python
print("\n=== MISSING VALUES SETELAH PERBAIKAN ===")

missing_after = df_dirty.isnull().sum()
print(missing_after)

total_missing_after = df_dirty.isnull().sum().sum()

print("\nTotal missing value setelah cleaning:",
      total_missing_after)

if total_missing_after == 0:
    print("STATUS: Semua missing value sudah berhasil ditangani.")
else:
    print("STATUS: Masih terdapat missing value.")
```

# 6. CEK DATA DUPLIKAT

```python
print("\n=== CEK DATA DUPLIKAT SEBELUM CLEANING ===")

dup_count = df_dirty.duplicated().sum()

print("Jumlah baris duplikat:", dup_count)


# Eksekusi penghapusan duplikat
before = len(df_dirty)

df_clean = df_dirty.drop_duplicates().copy()

after = len(df_clean)

print("\n=== HASIL PENGHAPUSAN DUPLIKAT ===")
print("Jumlah baris sebelum cleaning :", before)
print("Jumlah baris setelah cleaning  :", after)
print("Jumlah baris terhapus         :", before - after)
```

# 7. CEK ULANG DATA DUPLIKAT

```python
print("\n=== CEK DUPLIKAT SETELAH CLEANING ===")

dup_after = df_clean.duplicated().sum()

print("Jumlah duplikat setelah cleaning:", dup_after)

if dup_after == 0:
    print("STATUS: Tidak ada data duplikat.")
else:
    print("STATUS: Masih terdapat data duplikat.")
```

# 8. CEK DATA NUMERIK

```python
print("\n=== STATISTIK DATA NUMERIK ===")

numeric_columns = [
    'total_qty',
    'total_weight_gr',
    'total_returned_qty',
    'Total Diskon',
    'num_product_categories',
    'Ongkos Kirim Dibayar oleh Pembeli',
    'Estimasi Potongan Biaya Pengiriman',
    'Total Pembayaran',
    'Perkiraan Ongkos Kirim'
]

print(df_clean[numeric_columns].describe())
```

# 9. CEK NILAI NEGATIF DAN EKSEKUSI PERBAIKAN

```python
print("\n=== CEK NILAI NEGATIF ===")

for col in numeric_columns:

    negative_count = (df_clean[col] < 0).sum()

    print(f"{col}: {negative_count} nilai negatif")


# Karena kolom tersebut tidak semestinya memiliki nilai negatif,
# nilai negatif akan diubah menjadi NaN lalu ditangani dengan median.

for col in numeric_columns:

    negative_count = (df_clean[col] < 0).sum()

    if negative_count > 0:

        print(f"\nMemperbaiki {col}...")

        df_clean.loc[df_clean[col] < 0, col] = np.nan

        df_clean[col] = df_clean[col].fillna(
            df_clean[col].median()
        )

        print(f"{col} berhasil diperbaiki.")
```

# 10. CEK ULANG NILAI NEGATIF

```python
print("\n=== CEK NILAI NEGATIF SETELAH CLEANING ===")

total_negative = 0

for col in numeric_columns:

    negative_count = (df_clean[col] < 0).sum()

    print(f"{col}: {negative_count} nilai negatif")

    total_negative += negative_count


if total_negative == 0:
    print("\nSTATUS: Tidak ada nilai negatif.")
else:
    print("\nSTATUS: Masih terdapat nilai negatif.")
```

# 11. STANDARISASI DATA KATEGORIKAL

```python
print("\n=== NILAI METODE PEMBAYARAN SEBELUM STANDARISASI ===")

print(df_clean['Metode Pembayaran'].value_counts(dropna=False))


# Hilangkan spasi berlebih
df_clean['Metode Pembayaran'] = (
    df_clean['Metode Pembayaran']
    .astype(str)
    .str.strip()
)


# Samakan berbagai penulisan COD
payment_mapping = {
    'COD': 'COD (Bayar di Tempat)',
    'Cash On Delivery': 'COD (Bayar di Tempat)',
    'COD (Bayar di Tempat)': 'COD (Bayar di Tempat)',

    'OnlinePayment': 'Online Payment',
    'Online Payment': 'Online Payment'
}

df_clean['Metode Pembayaran'] = (
    df_clean['Metode Pembayaran']
    .replace(payment_mapping)
)


print("\n=== NILAI METODE PEMBAYARAN SETELAH STANDARISASI ===")
print(df_clean['Metode Pembayaran'].value_counts(dropna=False))
```

# 12. CEK ULANG KONSISTENSI KATEGORI

```python
print("\n=== CEK NILAI UNIK METODE PEMBAYARAN ===")
print(df_clean['Metode Pembayaran'].unique())

print("\n=== CEK NILAI UNIK STATUS PESANAN ===")
print(df_clean['Status Pesanan'].unique())

print("\n=== CEK NILAI UNIK PROVINSI ===")
print(df_clean['Provinsi'].unique())
```

# 13. CEK FORMAT TANGGAL

```python
print("\n=== ORDER DATE SEBELUM CLEANING ===")
print(df_clean[['order_id', 'Waktu Pesanan Dibuat', 'order_date']].tail(15))


# Konversi Waktu Pesanan Dibuat
df_clean['Waktu Pesanan Dibuat'] = pd.to_datetime(
    df_clean['Waktu Pesanan Dibuat'],
    errors='coerce'
)


# Konversi order_date
df_clean['order_date'] = pd.to_datetime(
    df_clean['order_date'],
    errors='coerce',
    dayfirst=True,
    format='mixed'
)


# Cari tanggal yang gagal dikonversi
invalid_date = df_clean['order_date'].isna().sum()

print("\nJumlah order_date tidak valid:", invalid_date)
```

# 14. PERBAIKAN TANGGAL TIDAK VALID

```python
if invalid_date > 0:

    print("\n=== MEMPERBAIKI TANGGAL TIDAK VALID ===")

    # Ambil tanggal dari Waktu Pesanan Dibuat
    # karena kolom tersebut berisi tanggal order yang valid
    df_clean.loc[
        df_clean['order_date'].isna(),
        'order_date'
    ] = df_clean.loc[
        df_clean['order_date'].isna(),
        'Waktu Pesanan Dibuat'
    ].dt.normalize()

    print("Tanggal tidak valid berhasil diperbaiki menggunakan")
    print("tanggal dari kolom Waktu Pesanan Dibuat.")

else:

    print("Tidak ditemukan tanggal tidak valid.")
```

# 15. CEK ULANG FORMAT TANGGAL

```python
print("\n=== CEK ORDER DATE SETELAH CLEANING ===")

invalid_date_after = df_clean['order_date'].isna().sum()

print("Jumlah order_date tidak valid:",
      invalid_date_after)

if invalid_date_after == 0:
    print("STATUS: Semua order_date valid.")
else:
    print("STATUS: Masih terdapat order_date yang tidak valid.")
```

# 16. KONVERSI TIPE DATA NUMERIK

```python
print("\n=== TIPE DATA SEBELUM STANDARISASI ===")
print(df_clean.dtypes)


# total_qty seharusnya berupa bilangan bulat
df_clean['total_qty'] = df_clean['total_qty'].round().astype('Int64')


# Kolom angka lainnya
integer_columns = [
    'total_weight_gr',
    'total_returned_qty',
    'Total Diskon',
    'num_product_categories',
    'Ongkos Kirim Dibayar oleh Pembeli',
    'Estimasi Potongan Biaya Pengiriman',
    'Total Pembayaran',
    'Perkiraan Ongkos Kirim'
]

for col in integer_columns:

    df_clean[col] = pd.to_numeric(
        df_clean[col],
        errors='coerce'
    ).round().astype('Int64')
```

# 17. CEK TIPE DATA SETELAH CLEANING

```python
print("\n=== TIPE DATA SETELAH CLEANING ===")
print(df_clean.dtypes)
```

# 18. CEK KOLOM TIDAK RELEVAN

```python
print("\n=== JUMLAH NILAI UNIK SETIAP KOLOM ===")
print(df_clean.nunique())


# Cari kolom yang hanya mempunyai satu nilai unik
single_value_cols = df_clean.nunique()[
    df_clean.nunique() == 1
].index.tolist()

print("\nKolom dengan hanya 1 nilai unik:")
print(single_value_cols)


# Eksekusi penghapusan
if len(single_value_cols) > 0:

    df_clean = df_clean.drop(
        columns=single_value_cols,
        errors='ignore'
    )

    print("\nKolom tidak relevan berhasil dihapus.")

else:

    print("\nTidak ada kolom yang perlu dihapus.")
```

# 19. CEK ULANG KOLOM

```python
print("\n=== KOLOM DATASET SETELAH CLEANING ===")
print(df_clean.columns.tolist())

print("\nJumlah kolom sekarang:", len(df_clean.columns))
```

# 20. DATA ENRICHMENT

```python
print("\n=== DATA ENRICHMENT ===")

# Jika total_returned_qty sudah terhapus pada proses sebelumnya,
# buat kembali dengan nilai 0
if 'total_returned_qty' not in df_clean.columns:
    df_clean['total_returned_qty'] = 0


# IsReturned
# 1 jika ada barang yang dikembalikan
# 0 jika tidak ada
df_clean['IsReturned'] = np.where(
    df_clean['total_returned_qty'] > 0,
    1,
    0
)


# IsCancelled
# 1 jika pesanan dibatalkan
# 0 jika tidak
df_clean['IsCancelled'] = np.where(
    df_clean['Status Pesanan'].str.strip().str.lower() == 'batal',
    1,
    0
)


# Net Quantity
df_clean['NetQuantity'] = (
    df_clean['total_qty'] -
    df_clean['total_returned_qty']
)


# Shipping Cost Ratio
df_clean['ShippingCostRatio'] = np.where(
    df_clean['Total Pembayaran'] > 0,
    df_clean['Perkiraan Ongkos Kirim'] /
    df_clean['Total Pembayaran'],
    0
)


print("Data enrichment berhasil ditambahkan.")
```
